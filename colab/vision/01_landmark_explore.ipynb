# Role 2 landmark exploration
Export per-frame measurements from the browser harness at the real 480 px input size. Use only a consented participant or generated media. Expected CSV columns: `ts_ms,mouth_aperture,yaw_deg,pitch_deg,roll_deg,landmark_quality`.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
csv_path = Path('/content/landmarks.csv')  # upload a consented clip's exported measurements
df = pd.read_csv(csv_path)
df.head()
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True)
axes[0].plot(df.ts_ms / 1000, df.mouth_aperture); axes[0].set_ylabel('Mouth aperture')
axes[1].plot(df.ts_ms / 1000, df.yaw_deg, label='yaw'); axes[1].plot(df.ts_ms / 1000, df.pitch_deg, label='pitch'); axes[1].legend()
axes[2].plot(df.ts_ms / 1000, df.landmark_quality); axes[2].set_ylabel('Landmark quality')
axes[2].set_xlabel('Seconds'); plt.tight_layout()
```
