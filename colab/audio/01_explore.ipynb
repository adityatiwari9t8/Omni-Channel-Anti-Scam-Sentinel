# AASIST clip inventory
Upload only consented, labeled evaluation clips. This notebook checks the role label file and WAV format before inference. The model expects mono float32 16 kHz audio and uses the first 64,600 samples.

```python
from pathlib import Path
import csv
import soundfile as sf
root = Path('../../eval/clips')
labels_path = root / 'labels/role1.csv'
rows = list(csv.DictReader(labels_path.open(encoding='utf-8', newline='')))
print(f'{len(rows)} labeled clips')
for row in rows:
    path = Path(row['file'])
    if not path.is_absolute(): path = Path('../..') / path
    info = sf.info(path)
    print(row['clip_id'], row['label'], info.samplerate, info.channels, round(info.duration, 2), 'sec')
    assert row['label'].lower() in {'real', 'fake'}
    assert info.samplerate == 16000 and info.channels == 1 and info.frames >= 64600
```
